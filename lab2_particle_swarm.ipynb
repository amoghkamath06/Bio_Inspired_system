```python
import numpy as np

n = 5          # Number of particles
iter = 50      # Number of iterations

w = 0.7        # Inertia weight
c1 = 1.5       # Personal learning factor
c2 = 1.5       # Global learning factor

# Initialize position and velocity
x = np.random.rand(n) * 10 - 5
v = np.zeros(n)

pbest = x.copy()
pbestval = x**2

gbestval = np.min(pbestval)
idx = np.argmin(pbestval)
gbest = pbest[idx]

for t in range(iter):
    for i in range(n):
        # Update velocity
        v[i] = w * v[i] + c1 * np.random.rand() * (pbest[i] - x[i]) + c2 * np.random.rand() * (gbest - x[i])

        # Update position
        x[i] = x[i] + v[i]

        # Calculate fitness
        f = x[i]**2

        # Update personal best
        if f < pbestval[i]:
            pbest[i] = x[i]
            pbestval[i] = f

        # Update global best
        if pbestval[i] < gbestval:
            gbest = pbest[i]
            gbestval = pbestval[i]

print(f'Best position = {gbest:.6f}')
print(f'Minimum function value = {gbestval:.6f}')
```
